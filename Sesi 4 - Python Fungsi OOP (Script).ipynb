## 1. Functions & Return Types

Fungsi adalah blok kode terisolasi yang menjalankan tugas spesifik dan dapat digunakan kembali (*reusable*). Fungsi merupakan implementasi standar dari prinsip **DRY (Don't Repeat Yourself)**.

### Komponen Fungsi
- **`def`**: Keyword untuk mendefinisikan fungsi.
- **Nama Fungsi**: Menggunakan *snake_case*.
- **Parameter & Argumen**: Input opsional ke dalam fungsi.
- **Return Type**: Output yang dikembalikan oleh fungsi.

```python
# 1. Fungsi Sederhana Tanpa Parameter & Tanpa Return (Return None secara default)
def sapa_peserta():
    print("Selamat datang di Bootcamp AI Engineer DigitalSkola!")

hasil_sapa = sapa_peserta()
print(f"Output dari fungsi: {hasil_sapa}")
print(f"Tipe output: {type(hasil_sapa)}\n")

# 2. Fungsi dengan Parameter dan Return Value
def tambah_angka(a: int, b: int):
    result = a + b
    return result

total = tambah_angka(15, 252)
print(f"Hasil Penjumlahan: {total}")
print(f"Tipe Data Output: {type(total)}")
```

## 2. Intro to OOP: Class & Instance

**Object-Oriented Programming (OOP)** adalah paradigma pemrograman yang mengorganisasi kode berdasarkan objek.
- **Class**: *Blueprint* atau cetakan untuk membuat objek.
- **Instance**: Objek nyata yang dibuat berdasarkan *blueprint* Class.
- **Constructor (`__init__`)**: Metode khusus untuk inisialisasi atribut objek.

```python
# Definisi Class
class Mobil:
    def __init__(self, merk: str, model: str, tahun: int):
        self.merk = merk
        self.model = model
        self.tahun = tahun

    def klakson(self):
        return f"{self.merk} {self.model}: Wok wok! 🚗"

# Membuat Instance Objek
mobil_a = Mobil(merk="Toyota", model="GR Yaris", tahun=2024)
mobil_b = Mobil(merk="Suzuki", model="Fronx", tahun=2025)

print(mobil_a.klakson())
print(f"Tahun Rilis: {mobil_b.tahun}")
```

## 3. Inheritance & Method Types

### A. Inheritance (Pewarisan)
Kemampuan *Child Class* untuk mewarisi atribut dan metode dari *Parent Class*.

### B. Tipe-Tipe Method
1. **Instance Method**: Terikat pada objek (`self`), dapat mengakses data instance.
2. **Class Method (`@classmethod`)**: Terikat pada class (`cls`), digunakan untuk manipulasi status class.
3. **Static Method (`@staticmethod`)**: Fungsi independen di dalam namespace class.

```python
# Parent Class
class Kendaraan:
    jumlah_kendaraan = 0  # Atribut Class

    def __init__(self, nama: str):
        self.nama = nama
        Kendaraan.jumlah_kendaraan += 1

    # Instance Method
    def info(self):
        return f"Nama Kendaraan: {self.nama}"

    # Class Method
    @classmethod
    def get_total_kendaraan(cls):
        return f"Total Kendaraan Terdaftar: {cls.jumlah_kendaraan}"

    # Static Method
    @staticmethod
    def is_valid_speed(speed: float) -> bool:
        return 0 <= speed <= 300

# Child Class (Inheritance)
class MobilListrik(Kendaraan):
    def __init__(self, nama: str, kapasitas_baterai: float):
        super().__init__(nama)
        self.kapasitas_baterai = kapasitas_baterai

    # Override Instance Method
    def info(self):
        return f"Mobil Listrik: {self.nama} | Baterai: {self.kapasitas_baterai} kWh"

# Eksekusi
ev1 = MobilListrik("Hyundai Ioniq 5", 72.6)
print(ev1.info())
print(Kendaraan.get_total_kendaraan())
print(f"Apakah kecepatan 150 km/jam valid? {Kendaraan.is_valid_speed(150)}")
```

## 4. Type Hinting & Pydantic Dasar

- **Type Hinting**: Menambahkan anotasi tipe data opsional di Python untuk meningkatkan *readability* dan dukungan IDE.
- **Pydantic**: Library data kustom untuk melakukan validasi tipe data secara *runtime*. Sangat penting pada arsitektur **AI Engineering** (misal: penanganan skema structured output LLM).

```python
from pydantic import BaseModel, Field, ValidationError

# Mendefinisikan Schema Menggunakan Pydantic
class UserProfile(BaseModel):
    id: int
    username: str
    email: str
    is_active: bool = True
    score: float = Field(gt=0, description="Skor harus lebih besar dari 0")

# Contoh 1: Data Valid & Otomatis Data Coercion
data_valid = {
    "id": "100",
    "username": "akmal_fauzan",
    "email": "akmal@searadata.com",
    "score": 88.5
}

user = UserProfile(**data_valid)
print("=== User Valid ===")
print(user)
print(f"Tipe 'id': {type(user.id)}\n")

# Contoh 2: Data Invalid (Trigger ValidationError)
try:
    data_invalid = {
        "id": 101,
        "username": "budi",
        "email": "invalid-email",
        "score": -10
    }
    bad_user = UserProfile(**data_invalid)
except ValidationError as e:
    print("=== Error Validasi Pydantic ===")
    print(e)
```

## 5. Latihan Studi Kasus: LLM Agent Response Schema

Skenario: Anda sedang membangun sistem AI Agent yang menerima instruksi pemrosesan teks dan harus memberikan output berupa struktur JSON terevaluasi menggunakan Pydantic.

```python
from pydantic import BaseModel, Field
from typing import List, Optional

# Structuring Schema AI Agent Output
class EntityExtraction(BaseModel):
    entity_name: str
    category: str
    confidence_score: float = Field(ge=0.0, le=1.0)

class AgentOutputSchema(BaseModel):
    query: str
    intent: str
    entities: List[EntityExtraction]
    summary: Optional[str] = None

# Contoh Output dari Model AI
raw_llm_response = {
    "query": "Berapa harga tiket pesawat ke Bali bulan depan?",
    "intent": "flight_inquiry",
    "entities": [
        {"entity_name": "Bali", "category": "Location", "confidence_score": 0.98},
        {"entity_name": "bulan depan", "category": "DateRange", "confidence_score": 0.85}
    ],
    "summary": "User menanyakan estimasi harga tiket penerbangan ke Bali."
}

# Validasi Output
parsed_agent_response = AgentOutputSchema(**raw_llm_response)

print("Status Parse AI Response:")
print(f"Intent terdeteksi: {parsed_agent_response.intent}")
print(f"Jumlah Entitas: {len(parsed_agent_response.entities)}")
for idx, entity in enumerate(parsed_agent_response.entities, 1):
    print(f"  {idx}. {entity.entity_name} ({entity.category}) - Confidence: {entity.confidence_score}")
```
